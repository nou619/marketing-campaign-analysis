```python
df['views_was_missing'] = df['views'].isna()

df['views'] = df.groupby(['channel', 'campaign_name'])['views'].transform(lambda x: x.fillna(x.median()))
```

```python
original = df[~df['views_was_missing']]['views']
imputed = df[df['views_was_missing']]['views']

print("Original median:", original.median())
```

```python

df['spend_was_missing'] = df['spend'].isna()
df['spend'] = df.groupby(['channel', 'campaign_name'])['spend'] \
                .transform(lambda x: x.fillna(x.median()))
original = df[~df['spend_was_missing']]['spend']
imputed = df[df['spend_was_missing']]['spend']

```

```python
#Spend missing values were imputed using the median per campaign and channel to preserve dataset completeness for analysis, because zero spend cannot be assumed. Revenue values were left as NaN because imputing them could introduce bias, so analysis on revenue was performed only on rows with actual recorded values.
```

```python

df['conversions_was_missing'] = df['conversions'].isna()
df['conversions'] = df.groupby(['channel', 'campaign_name'])['conversions'] \
                .transform(lambda x: x.fillna(x.median()))
original = df[~df['conversions_was_missing']]['conversions']
imputed = df[df['conversions_was_missing']]['conversions']
```

```python
df["rev_missing"] = df["clicks"].isna().astype(int)

channel_missing = (
    df.groupby("channel")["rev_missing"]
      .mean()
      .sort_values(ascending=False)
)

print(channel_missing)
```

```python
# Step 1: create missing flag BEFORE anything
df['clicks_missing'] = df['clicks'].isna()

# Step 2: fill missing clicks with channel + campaign median
df['clicks'] = df.groupby(['channel', 'campaign_name'])['clicks'] \
                 .transform(lambda x: x.fillna(x.median()))

# Step 3: enforce clicks = 0 where conversions = 0
df.loc[df["conversions"] == 0, "clicks"] = 0

# Step 4: recalc missing flag if you want to track imputed rows
df['clicks_imputed_flag'] = df['clicks_missing'] & (df["conversions"] != 0)

# Step 5: separate original vs imputed for median check
original = df.loc[~df['clicks_imputed_flag'], 'clicks']
imputed = df.loc[df['clicks_imputed_flag'], 'clicks']


```

```python
import numpy as np

# Flag missing revenue
df['revenue_was_missing'] = df['revenue'].isna()

# Handle structural zeros
df.loc[df['conversions'] == 0, 'revenue'] = 0
df.loc[df['conversions'] == 0, 'revenue_was_missing'] = False  # <- fix dtype

# Compute RPC (Revenue per Click)
df['rpc'] = df['revenue'] / df['clicks'].replace(0, np.nan)
df['rpc'] = df.groupby(['channel','campaign_name'])['rpc'].transform(lambda x: x.fillna(x.median()))

# Impute missing revenue
df['revenue'] = df['revenue'].fillna((df['rpc'] * df['clicks']).round())
df['revenue'] = df['revenue'].clip(lower=0)

# Check medians
original_revenue = df[~df['revenue_was_missing']]['revenue']
imputed_revenue = df[df['revenue_was_missing']]['revenue']


```
